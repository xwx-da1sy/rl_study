# 第 2 章补充：PMF 与 PDF

本课只回答一个问题：**已知随机变量的分布，怎样求某个结果或某段范围出现的概率？** 离散变量用 PMF，连续变量用 PDF。读完后应能区分“某点的概率”和“某点的概率密度”，并能各手算一个例子。


## 1. 先判断随机变量能取什么值

随机变量 $X$ 把一次随机试验的结果表示成数。掷骰子的点数只能是 $1,2,3,4,5,6$，属于**离散型**；在 $0$ 到 $1$ 秒之间的等待时间可以取无数个数值，属于**连续型**。

离散型变量适合逐个数值分配概率；连续型变量则要考虑一段区间占多少概率。下文先讨论纯离散或具有 PDF 的连续分布，暂不讨论两者混合的情况。


## 2. PMF：离散变量每个值的概率

PMF 是 *Probability Mass Function*，中文叫**概率质量函数**。对离散随机变量 $X$，定义为

$$
p_X(x)=P(X=x).
$$

所以 $p_X(3)$ 就是“$X$ 恰好等于 $3$”的概率。PMF 必须满足：每个值的概率非负，并且把所有可能取值的概率加起来等于 $1$：

$$
p_X(x)\geq 0,\qquad \sum_{x\in\mathcal{X}}p_X(x)=1.
$$

其中 $\mathcal{X}$ 是 $X$ 的可能取值集合。要计算多个值组成的事件，就把这些值的概率相加。


### 手算：公平六面骰

设 $X$ 为掷一次公平骰子的点数。六个结果等可能，因此 $p_X(1)=\cdots=p_X(6)=1/6$，其余取值的 PMF 为 $0$。例如：

$$
P(X=3)=p_X(3)=\frac{1}{6}.
$$

求偶数点数的概率时，把 $2,4,6$ 对应的质量相加：

$$
P(X\in\{2,4,6\})=p_X(2)+p_X(4)+p_X(6)=\frac{3}{6}=\frac{1}{2}.
$$

自检归一化：$6\times(1/6)=1$。可以把 PMF 想成分配给各个取值的“概率份额”。


## 3. PDF：连续变量在区间上的概率密度

PDF 是 *Probability Density Function*，中文叫**概率密度函数**。若连续随机变量 $X$ 有密度 $f_X(x)$，它满足

$$
f_X(x)\geq 0,\qquad \int_{-\infty}^{\infty}f_X(x)\,dx=1.
$$

区间概率等于曲线在该区间下的面积：

$$
P(a\leq X\leq b)=\int_a^b f_X(x)\,dx.
$$

这里 $f_X(x)$ 是**密度**，不是 $P(X=x)$。对于有 PDF 的连续分布，单个精确数值没有宽度，所以 $P(X=x)=0$。若区间很窄、密度变化也很小，可以用 $f_X(x)\,\Delta x$ 近似宽度为 $\Delta x$ 的区间概率；$\Delta x$ 不能省略。


### 手算：$[0,1]$ 上的均匀分布

设 $X$ 在 $0$ 到 $1$ 之间均匀取值，密度为

$$
f_X(x)=\begin{cases}1,&0\leq x\leq 1,\\0,&\text{其他位置}.\end{cases}
$$

整段曲线下面是宽 $1$、高 $1$ 的矩形，面积为 $1$。从 $0.2$ 到 $0.5$ 的概率是宽 $0.3$、高 $1$ 的矩形面积：

$$
P(0.2\leq X\leq 0.5)=\int_{0.2}^{0.5}1\,dx=0.3.
$$

此时 $f_X(0.3)=1$，但 $P(X=0.3)=0$。前者是密度高度，后者才是单点概率。对于这样的连续分布，区间端点是否写成等号也不影响结果。


## 4. 为什么 PDF 可以大于 $1$？

概率不能大于 $1$，但密度可以。设 $Y$ 只在 $[0,0.1]$ 上均匀取值。为了让总面积为 $1$，该区间上的密度必须是 $10$：

$$
f_Y(y)=10\quad(0\leq y\leq 0.1),\qquad\int_0^{0.1}10\,dy=1.
$$

例如 $P(0\leq Y\leq 0.02)=10\times0.02=0.2$。密度 $10$ 只表示概率集中在较窄的范围内，真正的区间概率仍是面积 $0.2$。


## 5. CDF：把两种表示连起来

不论 $X$ 离散还是连续，都可以用累积分布函数 CDF 表示“不超过 $x$”的概率：

$$
F_X(x)=P(X\leq x).
$$

离散变量把不大于 $x$ 的 PMF 相加；有 PDF 的连续变量则把密度从左向右积分：

$$
F_X(x)=\sum_{t\leq x}p_X(t)\quad\text{（离散）},\qquad F_X(x)=\int_{-\infty}^{x}f_X(t)\,dt\quad\text{（有 PDF 的连续分布）}.
$$

在上面的均匀分布例子中，$F_X(0.5)-F_X(0.2)=0.5-0.2=0.3$，与直接计算区间面积一致。


## 6. 求平均值时，也是“求和”与“积分”的区别

当期望存在时，离散变量的期望是“取值 $\times$ 对应概率”的总和；连续变量的期望是“取值 $\times$ 对应密度”在整条数轴上的积分：

$$
\mathbb{E}[X]=\sum_{x\in\mathcal{X}}x\,p_X(x)\quad\text{（离散）},\qquad\mathbb{E}[X]=\int_{-\infty}^{\infty}x\,f_X(x)\,dx\quad\text{（有 PDF 的连续分布）}.
$$

公平骰子的期望为 $(1+2+3+4+5+6)/6=3.5$；$[0,1]$ 均匀分布的期望为 $\int_0^1x\,dx=0.5$。在强化学习中计算未来回报的平均值，也会用到相同的加权思想。


## 7. 放到强化学习里看

如果智能体在一个状态只能选择“左、右”两个动作，策略可以直接给出离散动作概率。例如 $\pi(\text{左}\mid s)=0.7$、$\pi(\text{右}\mid s)=0.3$，这就是一个 PMF，两个动作的概率相加为 $1$。

如果动作是可连续调整的方向角或力度，策略可能给出动作的 PDF。此时某个精确动作值上的 $\pi(a\mid s)$ 是密度；要问动作落在某个范围的概率，需要对这个范围积分。

同样，有限状态环境的状态转移通常用离散概率描述；连续状态环境可能需要密度。现在只需先记住：**对离散结果求和，对连续范围积分**。


## 8. 对照与常见误解

| 问题 | PMF | PDF |
| --- | --- | --- |
| 描述什么变量？ | 离散变量 | 有密度的连续变量 |
| 函数在一个点的值是什么？ | 该点的概率 | 该点的密度高度 |
| 怎样求事件概率？ | 对事件包含的取值求和 | 对事件覆盖的区间积分 |
| 怎样检查总量？ | 所有概率之和等于 $1$ | 整体曲线下的面积等于 $1$ |

不要把 PDF 曲线的高度直接当作概率，也不要把“连续变量某个精确值的概率为 $0$”误解为这个值绝不可能出现。它表示的是：单独一个点没有面积，只有给定区间才有可计算的概率。


## 9. 自检：先自己算，再看答案

1. 公平骰子掷一次，$P(X\geq 5)$ 是多少？这一步用求和还是积分？
2. $X$ 在 $[0,1]$ 上均匀分布，$P(0.1\leq X\leq0.4)$ 和 $P(X=0.4)$ 分别是多少？
3. 某连续变量在 $[0,0.25]$ 上均匀分布。它在这段区间上的 PDF 高度是多少？高度大于 $1$ 有问题吗？
4. 某状态下的策略为 $\pi(a_1\mid s)=0.2$、$\pi(a_2\mid s)=0.5$、$\pi(a_3\mid s)=0.3$。它是 PMF 还是 PDF？


### 参考答案

1. $P(X\geq5)=p_X(5)+p_X(6)=2/6=1/3$，用求和。
2. 区间面积为 $0.4-0.1=0.3$；单点概率 $P(X=0.4)=0$。
3. 高度为 $1/0.25=4$；它是密度，总面积仍为 $4\times0.25=1$。
4. 三个离散动作的概率和为 $1$，因此是 PMF。
