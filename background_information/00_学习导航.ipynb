# 强化学习学习导航

基于已有的机器学习和深度学习学习进度，从小型决策环境开始，再逐步进入深度强化学习。每次只推进一个核心问题。


## 当前起点

- 机器学习：已覆盖 KNN、回归、决策树、集成学习、贝叶斯，并推进到 K-means；熟悉 NumPy 实现与模型评估。
- 深度学习：已完成 PyTorch 自动求导、训练循环、MLP/CNN，以及 Attention、ViT、Swin 实践；当前学习目标检测与 YOLO 基础。
- 强化学习：从 GridWorld 的状态转移和折扣回报开始。


## 学习路线

| 阶段 | 核心问题 | 动手产出 |
| --- | --- | --- |
| 1. 基础与环境 | 智能体、状态、动作、奖励、回合；MDP、折扣回报、探索与利用 | 手写 GridWorld，解释一条轨迹的状态转移和折扣回报 |
| 2. 表格方法 | 贝尔曼方程、价值迭代、Q-learning、SARSA、ε-greedy | NumPy 实现 Q-learning，对比随机策略并画奖励与成功率曲线 |
| 3. 深度价值方法 | 从 Q 表到神经网络；经验回放、目标网络、DQN | PyTorch 训练 CartPole DQN，报告多回合平均回报 |
| 4. 策略方法 | REINFORCE、优势函数、Actor-Critic，了解 PPO | 实现 REINFORCE，比较与 DQN 的训练稳定性 |

每阶段以 notebook 学习概念；需要动手实现时，再加入必要的短代码或独立实现。评估时使用独立回合，至少报告多回合平均回报；结果波动时检查不同随机种子。


## 现在开始

打开同目录下的 `01_GridWorld与折扣回报.ipynb`，先读懂轨迹与折扣回报，手算最后的练习，再用短代码单元核对结果。
