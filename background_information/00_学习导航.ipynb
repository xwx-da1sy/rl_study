以本地《深度强化学习》（drl_v1.pdf）的章节为参考，结合已有的机器学习和 PyTorch 基础推进。书仅供个人学习，不将原书或大段原文加入工程；笔记用自己的话解释，每次只解决一个核心问题。

- 深度学习：已完成 PyTorch 自动求导、训练循环、MLP/CNN，以及 Attention、ViT、Swin 实践；当前学习目标检测与 YOLO 基础。
- 强化学习：已开始 GridWorld、状态转移与折扣回报；当前先按原书复习第 1 章“深度学习基础”。


## 按书展开的学习路线

| 阶段 | 对应书中章节 | 重点与完成标志 |
| --- | --- | --- |
| 1. 基础知识 | 第 1–3 章 | 先复习第 1 章的线性模型、神经网络、反向传播与梯度下降，再学习蒙特卡洛和 MDP；能解释模型、损失、梯度与参数更新的关系。 |
| 2. 表格价值学习 | 第 4 章的 TD、Q-learning 与同策略/异策略；第 5 章表格 SARSA | 从贝尔曼关系走到 TD 更新，用 NumPy 在小环境比较随机策略、Q-learning 与 SARSA，解释探索的作用。 |
| 3. 深度价值学习 | 回到第 4 章 DQN；第 5–6 章按需深入 | 把 Q 表换成神经网络，理解经验回放和目标网络；用 PyTorch 在 CartPole 训练 DQN，记录独立评估回报。 |
| 4. 策略学习 | 第 7–8 章；第 9 章选读 | 推导策略梯度，依次理解 REINFORCE、基线和 Actor-Critic；在同一环境比较 REINFORCE 与 DQN 的回报和波动。 |
| 5. 后续专题 | 第 10–12 章；第 13–19 章选读 | 按兴趣学习连续控制、部分可观测、模仿学习、多智能体及应用；先选一个方向做小实验，不要求一次学完。 |

书的第 4 章先介绍 DQN，再讲 TD 和 Q-learning。这里先学 TD、表格 Q-learning 与 SARSA，再回看 DQN，便于理解神经网络究竟替代了什么。PPO 可在学完 Actor-Critic 后另作扩展，不作为这本书的必读章节。

每个主题采用「自己的话解释概念 → 手算小例子 → 必要时用短代码验证 → 自检」的节奏。实现算法时保留随机策略基线；评估使用独立回合，至少报告多回合平均回报，波动较大时检查不同随机种子。
